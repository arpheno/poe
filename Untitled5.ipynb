```python
import pandas as pd
from poe.ninja import retrieve_prices
from poe.valuation.framework.ninja_adapter import NinjaAdapter
from poe.valuation.framework.price_store import HashKeyPriceStore
from poe.valuation.framework.tft_adapter.adapter import TftAdapter
%matplotlib inline


import matplotlib.pyplot as plt
df= pd.read_csv('GuildStashHistory (5).csv')
```

```python
df
```

```python
mod=df.query('Action == "modified"')
```

```python
mod[['amount','Item']]=mod['Item'].str.split('×').apply(lambda x: pd.Series(x))
```

```python
prices = retrieve_prices()
adapter = NinjaAdapter()
tft_adapter = TftAdapter()
tft_valuations = tft_adapter.adapt()
valuations = adapter.adapt(prices)
price_store = HashKeyPriceStore(valuations)
price_store.add(tft_valuations)
```

```python
mod['Date'] = pd.to_datetime(mod['Date'])
```

```python
idx = mod.groupby(['Item'])['Date'].idxmax()
mod = mod.loc[idx]
```

```python
mod['value']=mod['Item'].map(lambda x:(price_store.query({'name':x.strip()})+[None])[0])
mod=mod.dropna()
mod['value']=mod['value'].apply(lambda x: x.estimate)
```

```python
mod['amount']=pd.to_numeric(mod['amount'])
mod['total']=mod.amount*mod.value
```

```python
to_plot=mod
```

```python
# identify items with a total value less than 10
other_items = to_plot[to_plot['total'] < 10]

# create a new row called "Other" that aggregates the total value of the small items
other_total = other_items['total'].sum()
other_row = pd.DataFrame({'Item': ['Other'], 'total': [other_total]})

# combine the "Other" row with the original DataFrame
to_plot = pd.concat([to_plot, other_row], ignore_index=True)

# remove the small items from the original DataFrame
to_plot = to_plot[~to_plot.isin(other_items)].dropna()
```

```python
mod=mod.sort_values('total')
# to_plot=mod[['Item','total']].query('Item.str.strip() != "Divine Orb"')
```

```python
# create a horizontal bar chart using Pandas
to_plot.plot(kind='barh', x='Item', y='total', figsize=(5, 15))
plt.title('Inventory Value')
plt.xlabel('Total')
plt.ylabel('Item')
plt.show()
```

```python
to_plot.set_index('Item')['total'].plot(kind='pie', figsize=(8, 8), autopct='%1.3f', legend=False)
plt.title('Inventory Value')
plt.ylabel('')
plt.show()
```

```python
# create a function to format the pie chart labels

from matplotlib.ticker import FuncFormatter
def format_func(val, pos):
    if val != 0:
        return f'{val:.0f}'
    else:
        return ''

# create a pie chart using Pandas
fig, ax = plt.subplots(figsize=(8, 8))
to_plot.set_index('Item')['total'].plot(kind='pie', autopct='', legend=False, labeldistance=None, textprops={'fontsize': 14}, pctdistance=0.8, wedgeprops={'linewidth': 2, 'edgecolor': 'white'}, ax=ax)
ax.set_title('Inventory Value', fontsize=16)
ax.set_ylabel('')
ax.xaxis.label.set_size(14)
ax.yaxis.label.set_size(14)
ax.xaxis.set_tick_params(labelsize=14)
ax.yaxis.set_tick_params(labelsize=14)
ax.xaxis.set_label_coords(0.5, -0.1)
ax.yaxis.set_label_coords(-0.1, 0.5)
ax.text(-1.5, 1.2, 'Value', fontsize=14, ha='center', va='center', transform=ax.transAxes)
ax.text(-1.5, 1.1, '($)', fontsize=14, ha='center', va='center', transform=ax.transAxes)
ax.text(-1.5, 1, f'Total: {to_plot["total"].sum():.0f}', fontsize=14, ha='center', va='center', transform=ax.transAxes)
ax.text(-1.5, -0.2, 'Figure 1: A pie chart showing the relative values of each inventory item.', fontsize=12, ha='left', va='center', transform=ax.transAxes)
ax.xaxis.set_major_formatter(FuncFormatter(format_func))
plt.show()
```
