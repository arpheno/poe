```python
from pydantic import BaseModel, validator
from typing import List

class Data(BaseModel):
    name: str
    divine: float
    chaos: int
    lowConfidence: bool
    ratio: int

class LoadData(BaseModel):
    timestamp: int
    data: List[Data]
    
    @validator('timestamp')
    def validate_timestamp(cls, v):
        if v < 0:
            raise ValueError("timestamp must be a positive integer")
        return v
```

```python

import requests

base_url = 'https://raw.githubusercontent.com/The-Forbidden-Trove/tft-data-prices/master/lsc/'
url = 'https://api.github.com/repos/The-Forbidden-Trove/tft-data-prices/contents/lsc'

response = requests.get(url)
files = response.json()

models = []
for file in files:
    file_name = file['name']
    file_url = base_url + file_name
    file_response = requests.get(file_url)
    file_data = file_response.json()
    model = LoadData.parse_obj(file_data)
    models.append(model)
```
