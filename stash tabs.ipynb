```python
import requests
from poe.poe_secrets import ssid as session_id
# Your account name and league
account_name = 'swozn'
league = 'necropolis'
# Your POESESSID for authentication
session_id='15d55b87a66f27111d813b0e3cfa5048'
# The URL for stash tab API
url = f'https://api.pathofexile.com/character-window/get-stash-items?accountName={account_name}&league={league}&tabs=1'

# Set up the session ID in the cookie for authentication
cookies = {'POESESSID': session_id}

def get_stash_tab_info():
    try:
        response = requests.get(url, cookies=cookies)
        response.raise_for_status()  # Raises an HTTPError if the response status code is 4XX/5XX
        data = response.json()
        
        if 'tabs' in data:
            for tab in data['tabs']:
                print(f"Tab Name: {tab.get('n')}, Tab ID: {tab.get('id')}, Tab Type: {tab.get('type')}")
        else:
            print("Could not find stash tab data in response.")
    except requests.exceptions.RequestException as e:
        print(f"An error occurred: {e}")

get_stash_tab_info()
```
