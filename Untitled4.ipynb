```python
import requests
from pydantic import BaseModel
from typing import List

class Item(BaseModel):
    name: str
    divine: float
    chaos: int
    lowConfidence: bool
    ratio: int

class Data(BaseModel):
    timestamp: int
    data: List[Item]

def load_data(file_url):
    response = requests.get(file_url)
    if response.status_code == 200:
        data = response.json()
        parsed_data = Data(**data)
        return parsed_data
    else:
        print(f"Failed to load data from the URL {file_url}")

url = "https://api.github.com/repos/The-Forbidden-Trove/tft-data-prices/contents/lsc"
response = requests.get(url)

if response.status_code == 200:
    files = response.json()
    file_urls = [file['download_url'] for file in files if file['name'].endswith('.json')]
    parsed_data_list = [load_data(file_url) for file_url in file_urls]
else:
    print("Failed to retrieve file names from the Github repository.")
```

```python
import requests

repo_url = "https://api.github.com/repos/The-Forbidden-Trove/tft-data-prices/contents/lsc"

def get_filenames(repo_url):
    response = requests.get(repo_url)
    contents = response.json()
    filenames = [content["name"] for content in contents if content["name"].endswith(".json")]
    return filenames

def load_data(filename):
    response = requests.get(f"https://raw.githubusercontent.com/The-Forbidden-Trove/tft-data-prices/master/lsc/{filename}")
    return response.json()

filenames = get_filenames(repo_url)
models = []
for filename in filenames:
    data = load_data(filename)
    parsed_data = Data.parse_obj(data)
    models.append(parsed_data)
```
